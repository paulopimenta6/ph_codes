```python
import spacy
#spacy.__version__
```

```python
nlp_pt = spacy.load('pt_core_news_sm') #tentando carregar o modelo pt (portuguese) do spacy em um objeto python
nlp_en = spacy.load('en_core_web_sm') #tentando carregar o modelo en (english) do spacy em um objeto python
```

```python
doc_pt = nlp_pt('Paulo está estudando ciência de dados e estatística')
doc_en = nlp_en('I am learning how to build chatbots')
for token in doc_pt:
    print(token.text, " - ", token.pos_)
print("--------------------------------------------------------------")
for token in doc_en:
    print(token.text, " - ", token.pos_)
print("--------------------------------------------------------------")
doc_en_london = nlp_en('I am going to London next week for a meeting.')
for token in doc_en_london:
    print((token.text, token.pos_)) 
```

```python
doc_pt_paulo_mestrado = nlp_pt("Paulo faz mestrado na Faculdade de Medicina da Universidade de São Paulo")
for token in doc_pt_paulo_mestrado:
    print((token.text, token.pos_))
```

```python
print("\"" + str(doc_en) + "\"")
print("Text - ", "Lemma - ", "POS - ", "TAG - ", "Dep - ", "Shape - ", "Alpha - ", "Stop")
for token in doc_en:
    print((token.text, token.lemma_, token.pos_, token.tag_, token.dep_, token.shape_, token.is_alpha, token.is_stop))
```

```python
print("componentes do spaCy3.x em ingles -> " + str(nlp_en.pipe_names))
print("componentes do spaCy3.x em portugues -> " + str(nlp_pt.pipe_names))
```

```python
doc = nlp_pt("Os gatos estavam correndo pelos jardins.")

for token in doc:
    print(token.text, "->", token.lemma_)
```

```python
doc = nlp_en("blazing")
print("O tipo do doc é um próprio doc que contem vários tokens: " + str(type(doc)))
print("Os métodos do objeto são:")
for m in dir(doc[0]):
    if not m.startswith("_"):
        print("método -> " + str(m))
```

```python
doc = nlp_en("blazing")
for token in doc:
    print(token.text, " -> ", token.lemma_)
```

```python
from nltk.stem.porter import *
from nltk.stem.snowball import SnowballStemmer

porter_stemmer = PorterStemmer()
snowball_stemmer_en = SnowballStemmer("english") 
snowball_stemmer_pt = SnowballStemmer("portuguese") 
print(porter_stemmer.stem("fastest")) #Com PorterStremmer
print(snowball_stemmer_en.stem("fastest"))
print(snowball_stemmer_pt.stem("rapidamente"))
```

```python
my_string = "Google has its headquarters in Mountain View, California having revenue amounted to 109.65 billion US dollars"
doc = nlp_en(my_string)
for ent in doc.ents:
    print((ent.text, ent.label_))
```

```python
doc1 = nlp_en('Book me a flight from Bangalore to Goa')
doc2 = nlp_pt('Reserve uma passagem de Bangalore para Goa')

blr_en, goa_en = doc1[5], doc1[7]
print(doc1)
print("Palavra principal -> ", blr_en)
print("Palavra principal -> ", goa_en)
print(list(blr_en.ancestors))
print(list(goa_en.ancestors))
print("-----------------------------------------------------------")
print(doc2)
blr_pt, goa_pt = doc2[4], doc2[6]
print("Palavra principal -> ", blr_pt)
print("Palavra principal -> ", goa_pt)
print(list(blr_pt.ancestors))
print(list(goa_pt.ancestors))
```
