```python
import spacy
from spacy import displacy
#spacy.__version__
```

```python
nlp_pt = spacy.load('pt_core_news_lg') #tentando carregar o modelo pt (portuguese) do spacy em um objeto python
nlp_en = spacy.load('en_core_web_lg') #tentando carregar o modelo en (english) do spacy em um objeto python
```

```python
doc_pt = nlp_pt('Paulo está estudando ciência de dados e estatística')
doc_en = nlp_en('I am learning how to build chatbots')
for token in doc_pt:
    print(token.text, " - ", token.pos_)
print("--------------------------------------------------------------")
for token in doc_en:
    print(token.text, " - ", token.pos_)
print("--------------------------------------------------------------")
doc_en_london = nlp_en('I am going to London next week for a meeting.')
for token in doc_en_london:
    print((token.text, token.pos_)) 
```

```python
doc1 = nlp_en('Book me a flight from Bangalore to Goa')
doc2 = nlp_pt('Reserve uma passagem de Bangalore para Goa')

blr_en, goa_en = doc1[5], doc1[7]
print(doc1)
print("Palavra principal -> ", blr_en)
print("Palavra principal -> ", goa_en)
print(list(blr_en.ancestors))
print(list(goa_en.ancestors))
print("-----------------------------------------------------------")
print(doc2)
blr_pt, goa_pt = doc2[4], doc2[6]
print("Palavra principal -> ", blr_pt)
print("Palavra principal -> ", goa_pt)
print(list(blr_pt.ancestors))
print(list(goa_pt.ancestors))
```

```python
doc_nlp_en = nlp_en("Book a table at the restaurant and the taxi to the hotel")

tasks = doc_nlp_en[2], doc_nlp_en[8]
tasks_target = doc_nlp_en[5], doc_nlp_en[11]

print(tasks)
print(tasks_target)

print("-" * 69)

for target in tasks_target:
    print(
        f"Ancestrais de {target} -> "
        f"{list(target.ancestors)}"
    )

print("-" * 69)

for target in tasks_target:
    for task in target.ancestors:
        if task in tasks:
            print(f"Booking of {task} belongs to {target}")
            break

print("-" * 69)

print(list(tasks_target[0].children)) #filhos sintáticos de uma palavra
```

```python
#import spacy

#nlp_en = spacy.load("en_core_web_lg")

# Buscando semelhanças
doc_en = nlp_en("How are you doing today?")
for token in doc_en:
    print(token.text, token.vector[:5])

hello_doc = nlp_en("hello")
hi_doc = nlp_en("hi")
hella_doc = nlp_en("hella")
# Buscando semelhanças com pequenas palavras
print("hello x hi:", hello_doc.similarity(hi_doc))
print("hello x hella:", hello_doc.similarity(hella_doc))
print("------------------------------------------------------------------")
frase_pt_1 = nlp_pt("Quando você irá ler o livro sobre Leonardo Da Vinci?")
frase_pt_2 = nlp_pt("O livro sobre Leonardo Da Vinci será lido por você?")
frase_pt_3 = nlp_pt("Hoje está fazendo muito calor ou frio em São Paulo?")
print("O nível de similaridade da fase 1 para frase 2 é: ", frase_pt_1.similarity(frase_pt_2))
print("O nível de similaridade da fase 1 para frase 3 é: ", frase_pt_1.similarity(frase_pt_3))
print("O nível de similaridade da fase 2 para frase 3 é: ", frase_pt_2.similarity(frase_pt_3))
```
